## Phase 6 — Step 3: AI Context Builder

Copy this after Step 1 and Step 2

```python
# ============================================================
# STEP 3 — AI CONTEXT BUILDER
# ============================================================

from typing import Any, Dict


def build_ai_context(
    product_request: Dict[str, Any],
    persona_profiles: Dict[str, Dict[str, Any]]
) -> Dict[str, Any]:
    """
    Build structured AI context from the employee-selected,
    data-driven Gen Z persona.
    """

    target_persona = product_request.get("target_persona")
    number_of_concepts = product_request.get("number_of_concepts", 3)

    if target_persona not in persona_profiles:
        valid_personas = ", ".join(persona_profiles.keys())
        raise ValueError(
            f"Unknown persona: {target_persona}. "
            f"Valid personas are: {valid_personas}"
        )

    if not isinstance(number_of_concepts, int) or number_of_concepts < 1:
        raise ValueError("number_of_concepts must be a positive integer.")

    persona = persona_profiles[target_persona]

    ai_context = {
        "target_persona": target_persona,
        "source_customer_cluster": persona["cluster"],
        "persona_profile": persona["profile"],
        "persona_characteristics": persona["characteristics"],
        "product_opportunity": persona["product_opportunity"],
        "number_of_concepts": number_of_concepts,
        "generation_method": (
            "The LLM independently selects the product category, "
            "reward architecture, pricing, eligibility, features, "
            "benefits, and value proposition."
        ),
        "important_scope": [
            "The persona was identified through unsupervised customer clustering.",
            "The persona information is context for ideation, not a validation result.",
            "Do not copy the product opportunity wording directly.",
            "Do not assume that every characteristic must become a card feature.",
            "Generated products will be independently evaluated after generation."
        ]
    }

    return ai_context
```

# Step 4 — Build the OpenAI prompt

This prompt gives the LLM customer intelligence but does not give it validation thresholds.

```python
# ============================================================
# STEP 4 — PRODUCT GENERATION PROMPT BUILDER
# ============================================================

import json
from typing import Any, Dict


def build_product_generation_prompt(
    ai_context: Dict[str, Any]
) -> str:
    """
    Convert the structured persona context into a product-generation prompt.
    """

    characteristics_text = "\n".join(
        f"- {item}"
        for item in ai_context["persona_characteristics"]
    )

    profile_json = json.dumps(
        ai_context["persona_profile"],
        indent=2
    )

    number_of_concepts = ai_context["number_of_concepts"]

    prompt = f"""
You are a senior Canadian credit card product innovation specialist.

OBJECTIVE

Generate {number_of_concepts} distinct and innovative consumer credit card
concepts for the selected Gen Z persona.

The persona was identified from an unsupervised customer-segmentation model.
Use the customer evidence below as product-development context.

TARGET PERSONA

{ai_context["target_persona"]}

SOURCE CUSTOMER CLUSTER

Cluster {ai_context["source_customer_cluster"]}

PERSONA CHARACTERISTICS

{characteristics_text}

CUSTOMER CLUSTER PROFILE

{profile_json}

GENERAL PRODUCT OPPORTUNITY

{ai_context["product_opportunity"]}

IDEATION RESPONSIBILITY

You must independently determine:

- Product concept
- Product category
- Annual fee
- Minimum income requirement
- Minimum credit score
- Purchase interest rate
- Foreign transaction fee
- Base reward rate
- Reward program type
- Reward architecture
- Welcome offer
- Digital features
- Reward features
- Subscription features
- Lifestyle features
- Sustainability features
- Financial wellness features
- Security and protection features
- Premium benefits
- Customer value proposition
- Business justification

CREATIVITY REQUIREMENTS

- Generate meaningfully different concepts.
- Do not generate minor variations of the same card.
- Each concept must use a different primary value proposition.
- Use different combinations of pricing, rewards, benefits, and features.
- Keep every concept plausible for the Canadian consumer credit card market.
- Do not merely repeat the persona description.
- Do not copy existing card names.
- Do not claim the concept is profitable without financial evidence.
- Do not evaluate similarity or predict product identity.
- Downstream analytical models will perform independent validation.

OUTPUT REQUIREMENTS

Return only valid JSON.

Use this exact structure:

{{
  "target_persona": "{ai_context["target_persona"]}",
  "concepts": [
    {{
      "product_name": "string",
      "product_category": "string",
      "concept_theme": "string",
      "annual_fee_cad": 0.0,
      "minimum_income_requirement": 0,
      "minimum_credit_score": 0,
      "purchase_interest_rate": 0.0,
      "foreign_transaction_fee_percent": 0.0,
      "base_reward_rate": 0.0,
      "welcome_bonus_value_cad": 0.0,
      "reward_program_type": "string",
      "reward_architecture": "string",

      "travel_rewards": 0,
      "airport_lounge_access": 0,
      "travel_insurance": 0,
      "trip_cancellation_insurance": 0,
      "baggage_insurance": 0,
      "rental_car_insurance": 0,
      "no_foreign_transaction_fee": 0,
      "hotel_rewards": 0,
      "airline_rewards": 0,

      "cashback_rewards": 0,
      "grocery_rewards": 0,
      "dining_rewards": 0,
      "fuel_rewards": 0,
      "transit_rewards": 0,
      "shopping_rewards": 0,

      "mobile_wallet_support": 0,
      "apple_pay_support": 0,
      "google_pay_support": 0,
      "samsung_pay_support": 0,
      "virtual_card_support": 0,
      "instant_card_issuance": 0,
      "digital_first_card": 0,

      "subscription_rewards": 0,
      "streaming_rewards": 0,
      "netflix_benefit": 0,
      "spotify_benefit": 0,
      "disney_plus_benefit": 0,
      "amazon_prime_benefit": 0,
      "youtube_premium_benefit": 0,
      "food_delivery_benefit": 0,
      "uber_benefit": 0,
      "instacart_benefit": 0,
      "doordash_benefit": 0,
      "skip_benefit": 0,

      "ride_share_rewards": 0,
      "gaming_rewards": 0,
      "entertainment_rewards": 0,

      "sustainability_rewards": 0,
      "carbon_tracking": 0,
      "green_rewards_program": 0,
      "sustainable_purchase_rewards": 0,

      "credit_score_monitoring": 0,
      "budgeting_tools": 0,
      "spending_insights": 0,
      "installment_payment_option": 0,
      "debt_management_support": 0,

      "fraud_alerts": 0,
      "identity_protection": 0,
      "purchase_protection": 0,
      "zero_liability": 0,

      "concierge_service": 0,
      "premium_hotel_benefits": 0,
      "luxury_travel_credits": 0,
      "exclusive_event_access": 0,

      "customer_value_proposition": "string",
      "business_justification": "string"
    }}
  ]
}}

BINARY FEATURE RULE

For every binary product feature:

- 1 means the feature is included.
- 0 means the feature is not included.
- Use integers only.
- Include every required field.
"""

    return prompt.strip()
```

```python
binary_columns = sorted({
    column
    for columns in feature_groups.values()
    for column in columns
})

missing_columns = [
    column
    for column in binary_columns
    if column not in generated_df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing binary feature columns: {missing_columns}"
    )

invalid_binary_values = {}

for column in binary_columns:
    invalid_values = generated_df.loc[
        ~generated_df[column].isin([0, 1]),
        column
    ].unique()

    if len(invalid_values) > 0:
        invalid_binary_values[column] = invalid_values.tolist()

if invalid_binary_values:
    raise ValueError(
        f"Invalid binary values found: {invalid_binary_values}"
    )

print("All required binary features are valid.")
```

# Phase 6: Persona-Led AI Product Innovation Engine

## Purpose

This phase creates an AI-assisted decision-support workflow for bank product employees. The employee selects a Gen Z persona directly, adds product preferences and commercial guardrails, and receives structured credit card concepts.

## Updated Workflow

1. The employee selects a target Gen Z persona.
2. The employee selects reward, feature, and annual-fee preferences.
3. The employee provides commercial guardrails, including profitability priority, reward-cost tolerance, credit-risk tolerance, and welcome-bonus budget.
4. The AI Context Builder retrieves the selected persona's needs.
5. It adds relevant market and product-cluster evidence as supporting context.
6. The Dynamic Prompt Builder asks the LLM to create realistic product concepts.
7. Each concept includes product specifications, likely revenue sources, major cost drivers, risk controls, profitability rationale, and business justification.
8. The generated concepts are saved for downstream feature engineering, similarity analysis, classification, and evaluation.

## Important Scope Decision

The market-gap result is no longer treated as proof of demand or profitability. A low number of existing products may reflect weak demand, high costs, risk, regulation, or low commercial attractiveness. Therefore, the market gap is used only as supporting evidence.

The system also does not calculate actual profit at this stage. The current synthetic product dataset does not contain the detailed revenue and cost information required for a financial forecast, such as interchange revenue, interest income, reward redemption cost, acquisition cost, servicing cost, fraud loss, or credit loss.

Instead, the LLM follows profitability and risk guardrails and provides a qualitative commercial rationale. A future version could calculate a profitability score if reliable bank revenue and cost assumptions become available.

## Business Value

The updated workflow keeps the product concept centered on the selected Gen Z persona while preventing the LLM from proposing expensive rewards, benefits, or acquisition offers without explaining how they may be commercially supported.
